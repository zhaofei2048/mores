```python
import numpy as np
import scipy.integrate as integrate
```

```python
fx = lambda x: np.array([[1, 2], [3, 4]]) @ np.array([[np.sin(x), 2*np.sin(x)], [x, x**2]]) @ np.array([[np.sin(x), 2*np.sin(x)], [x, x**2]])
integrate.quad_vec(fx, 0, 1)
```
