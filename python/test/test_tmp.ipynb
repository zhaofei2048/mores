```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
theta_is = np.linspace(1, 70, 70)

N = len(theta_is)
r = np.zeros(N)
r2 = np.zeros(N)
for i in range(N):
    theta_i = theta_is[i]
    geom = (theta_i, 180, theta_i, 0)
    ths, phis, thi, phi = np.deg2rad(geom)
    s = np.sin(thi)
    cs = np.cos(thi)
    ss = np.sin(ths)
    css = np.cos(ths)
    sf = np.sin(phis - phi)
    csf = np.cos(phis - phi)
    # sf = 0
    # csf = 1

    # Ulaby 2014, eq. 11.51, p474
    p11 = ss**2 * s**2 + css**2 * cs**2 * csf**2 + 2 * ss *s *css *cs * csf
    # Chen 2010
    pvv = 0.5*(2 * (1-cs**2)*(1-css**2) + cs**2 * css**2 * (2 * csf**2) + 4 * css * cs * np.sqrt((1-cs**2)*(1-css**2))*csf)
    # p11 = ss**2 * s**2 + css**2 * cs**2 + 2 * ss *s *css *cs
    r[i] = p11
    r2[i] = pvv

fig, ax = plt.subplots()
# ax.plot(theta_is, 10*np.log10(r))
ax.plot(theta_is, r, 'r-', label='Ulaby')
ax.plot(theta_is, r2, 'g--', label='chen')
plt.show()
```
