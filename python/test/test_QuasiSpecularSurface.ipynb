```python
import sys
sys.path.append('..')

import numpy as np
from QuasiSpecularSurface import QuasiSpecularSurface
import matplotlib.pyplot as plt
```

```python
epsr = 15-4j
kdel = 0.40868977928057787
surface = QuasiSpecularSurface(epsr, kdel)
```

```python

theta_is = np.linspace(1, 70, 70)

N = len(theta_is)
Rv = np.zeros(N)
Rh = np.zeros(N)
for i in range(N):
    theta_i = theta_is[i]
    M= surface.M_coh_R(theta_i)
    Rv[i] = 10*np.log10(4*np.pi*M[0,0])
    Rh[i] = 10*np.log10(4*np.pi*M[1,1])

fig, ax = plt.subplots()
ax.plot(theta_is, Rv, 'b', label='v')
ax.plot(theta_is, Rh, 'g--', label='h')
plt.legend()
plt.show()
```
