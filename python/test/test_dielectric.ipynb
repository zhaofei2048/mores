```python
# Compare with SMRT
import sys
sys.path.append('E:/exps/lunar_scattering/lunar_regolith_scattering/mores/python')
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

from dieletric_models import diel_soil
```

```python
mv  = np.linspace(0.01, 0.6, 50)
N = len(mv)
epsrs = np.zeros(N, dtype='complex')


for i in range(N):
    epsrs[i] = diel_soil(18e9, mv[i], 30.6e-2, 13.5e-2, 23)
```

```python
# mpl.rcParams['lines.linewidth'] = 2
plt.style.use('classic')
# mpl.rcParams['lines.linestyle'] = '--'
fig, ax = plt.subplots()
ax.plot(mv, np.real(epsrs))
ax.plot(mv, np.imag(epsrs))
ax.grid()
ax.set_xlim(0.0, 0.5)
ax.set_ylim(0, 30)
plt.savefig("../../images/diel_soil.png", dpi=300)
plt.show()
```
