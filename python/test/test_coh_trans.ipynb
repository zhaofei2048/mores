```python
import sys
sys.path.append('..')
import numpy as np
from RoughSurface import RoughSurface
import matplotlib.pyplot as plt
from postprocessing import Mueller2sigma0VH
```

```python
rs = RoughSurface(1/(2.7-0.2j), 0.3)
```

```python
rs.critical_angle()
```

```python
theta_is = np.linspace(1, 70, 70)
Num = len(theta_is)
sigma0_VVs = np.zeros(Num)
sigma0_HHs = np.zeros(Num)
for i in range(Num):
    theta_i = theta_is[i]
    Mue = rs.M_coh_transmit(theta_i)
    sigma_VV, _, _, sigma_HH = Mueller2sigma0VH(Mue)
    sigma0_VVs[i] = sigma_VV
    sigma0_HHs[i] = sigma_HH

sigma0_VV_dB = 10*np.log10(sigma0_VVs.real)
sigma0_HH_dB = 10*np.log10(sigma0_HHs.real)
sigma0_VV_dB = (sigma0_VVs)
sigma0_HH_dB = (sigma0_HHs)
plt.figure(figsize=(6, 4))
plt.plot(theta_is, sigma0_VV_dB, label='VV')
plt.plot(theta_is, sigma0_HH_dB, label='HH')
plt.xlabel('Incident angle (deg)')
plt.ylabel('$\sigma^0$ (dB)')
plt.legend()
plt.show()
```
