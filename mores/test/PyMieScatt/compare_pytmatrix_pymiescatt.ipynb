```python
import numpy as np
m = 1.77-0.63j
wavelength = 375
diameter = 30
radius = diameter/2
x = np.pi * diameter / wavelength
# epsr = np.conj(m**2)
```

```python
from pytmatrix import tmatrix, tmatrix_aux
from pytmatrix.scatter import sca_xsect, ext_xsect, ssa
from pytmatrix.radar import radar_xsect
scatterer = tmatrix.Scatterer(radius=radius, wavelength=wavelength, m=np.conj(m), axis_ratio=1.0)

# A = np.pi * radius**2   # the geometry cross section of the sphere
cross_sections = {'Qext': ext_xsect(scatterer),
 'Qsca': sca_xsect(scatterer)}
scatterer.set_geometry(geom=tmatrix_aux.geom_vert_back)
Z = scatterer.get_Z()
S = scatterer.get_S()
sigma_back = radar_xsect(scatterer=scatterer)

print(f"S: {S}")
print(f"Z: {Z}")
print(f"sigma_back: {sigma_back}")
print(cross_sections)
```
