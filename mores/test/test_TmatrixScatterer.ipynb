```python
import sys
sys.path.append('..')
import numpy as np
import matplotlib.pyplot as plt
import scipy.constants as C

from TmatrixScatterer import TmatrixScatterer
```

```python
# single particle size distribution
f = 4e9
epsr_background = 2.7-0.01j
epsr_particle = 6-0.1j
Lambda0 = C.speed_of_light / f
Lambda = Lambda0 / np.sqrt(np.real(epsr_background))

# particles
axis_ratio = 1
particle_size = [0.2*Lambda, 0.001]  # (radius, fs) or (0, Dmax, particle_size_distribution_func)
particle_orientation = 0    # scalar or (alpha, beta)
layer = TmatrixScatterer(f, epsr_background, epsr_particle, particle_size, particle_orientation, particle_shape=(axis_ratio, 'SPHEROID'), thickness=None)
layer.single_scattering_albedo()
```

```python
from pytmatrix.psd import ExponentialPSD
psd_func = ExponentialPSD(N0=1.0, Lambda=1.0, D_max=None)
# multiple particle sizes distribution
f = 4e9
epsr_background = 2.7-0.01j
epsr_particle = 6-0.1j
Lambda0 = C.speed_of_light / f
Lambda = Lambda0 / np.sqrt(np.real(epsr_background))

# particles
axis_ratio = 1
particle_size = (0, 0.2*Lambda, psd_func)  # (radius, fs) or (0, Dmax, particle_size_distribution_func)
particle_orientation = 0    # scalar or (alpha, beta)
layer = TmatrixScatterer(f, epsr_background, epsr_particle, particle_size, particle_orientation, particle_shape=(axis_ratio, 'SPHEROID'), thickness=None)
layer.single_scattering_albedo()
```
