```python
import numpy as np
import scipy.constants as sci_const
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(['science', 'notebook'])

import sys
sys.path.append('../../../')
%load_ext autoreload
%autoreload 2

# import miepython
from mores.misc.lunar_rocks_psd import lunar_psd_vol, lunar_psd_area, LunarRocksPSD
```

```python
k = np.linspace(1e-3, 0.1, 5)
D = np.logspace(-6, 1, 1000)
psd_vols = np.zeros((len(k), len(D)))
for i in range(len(k)):
    psd_obj = LunarRocksPSD(k[i])
    psd_vols[i, :] = psd_obj.lunar_psd_vol(D)

fig, ax = plt.subplots(1, 1)
ax.set_xscale('log')
ax.set_yscale('log')
for i in range(len(k)):
    ax.plot(D, psd_vols[i, :], label=f'k={k[i]}')
plt.legend()
ax.set_xlim(1e-3, 1e2)
ax.set_ylim(1e2, 1e9)
```

```python
k = [0.0025, 0.0031, 0.0117, 0.05, 0.097, 0.1]
D = np.logspace(-6, 1, 1000)
psd_vols = np.zeros((len(k), len(D)))
for i in range(len(k)):
    psd_vols[i, :] = lunar_psd_area(D, k=k[i])

fig, ax = plt.subplots(1, 1)
ax.set_xscale('log')
ax.set_yscale('log')
for i in range(len(k)):
    ax.plot(D, psd_vols[i, :], label=f'k={k[i]}')
plt.legend()
ax.set_xlim(1e-4, 1e3)
ax.set_ylim(1e-8, 1e6)
```
