```python
# Ulaby 2014, Fig. 2-17
theta_i = np.linspace(0, 90, 90)
er_water = 81
er_wetsoil = 25
er_drysoil = 3

rv_water, rh_water, _, _ = fresnel_coefficients(theta_i, er_water)
rv_wetsoil, rh_wetsoil, _, _ = fresnel_coefficients(theta_i, er_wetsoil)
rv_drysoil, rh_drysoil, _, _ = fresnel_coefficients(theta_i, er_drysoil)

plt.figure(figsize=(5, 5))
plt.plot(theta_i, np.abs(rv_water), 'b', label='abs(rv) (water)')
plt.plot(theta_i, np.abs(rh_water), 'b--', label='abs(rh) (water)')
plt.plot(theta_i, np.abs(rv_wetsoil), 'r', label='abs(rv) (wet soil)')
plt.plot(theta_i, np.abs(rh_wetsoil), 'r--', label='abs(rh) (wet soil)')
plt.plot(theta_i, np.abs(rv_drysoil), 'g', label='abs(rv) (dry soil)')
plt.plot(theta_i, np.abs(rh_drysoil), 'g--', label='abs(rh) (dry soil)')
plt.grid()
plt.legend()
plt.xlabel('incident angle (degree)')
plt.ylabel(r'|$\rho_v$| or |$\rho_h$|')
# plt.savefig('../images/Ulaby2014_fig_2_17.png')
plt.show()
```
