```python
import numpy as np
```

```python
f = 2.38e9
thickness = 1.0
epsr_background = 2.7 - 0.002j
epsr_particle = 6.0 - 0.2j
wavelength = 3e8 / f
fs = 0.001

# Default and derived
D = 0.1*wavelength
radius = D * 0.5
geometric_cross_section = np.pi * radius**2
Dmax = D
```

```python
from mores.layer.MieSphere import MieSphere
from mores.layer.TmatrixScatterer import TmatrixScatterer
from mores.misc.lunar_rocks_psd import LunarRocksPSD
```

```python
psd_obj = LunarRocksPSD(k=fs)
particle_size = (0, Dmax, psd_obj.lunar_psd_vol)
```

```python
miesphere_layer = MieSphere(f=f, thickness=thickness, epsr_background=epsr_background, epsr_particle=epsr_particle, 
                            particle_size=particle_size, num_points=100)
```

```python
tmatrix_layer = TmatrixScatterer(f=f, thickness=thickness, epsr_background=epsr_background, epsr_particle=epsr_particle, particle_size=particle_size, particle_shape=(1.0, 'SPHEROID'))
```
