```python
"""Test Hagfors backscattering model"""
import sys
sys.path.append('../../../')
%load_ext autoreload
%autoreload 2

import numpy as np
import scipy.constants as sci_const
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(['science', 'notebook'])

from mores.interface.Hagfors_backscattering import sigma0_pp_Hagfors
```

```python
epsr = 2.7 - 0.003j
C = [10, 50, 100, 1000]
theta_is = np.linspace(0, 70, 70)
sigma0_pps = np.zeros((len(C), len(theta_is)))
for i in range(len(C)):
    sigma0_pps[i] = sigma0_pp_Hagfors(theta_is, epsr, C[i])
sigma0_pps_dB = 10 * np.log10(sigma0_pps)

for i in range(len(C)):
    plt.plot(theta_is, sigma0_pps_dB[i], label=f'$C_H$={C[i]}')
plt.legend()
plt.grid()
plt.xlabel('Incidence angle (deg)')
plt.ylabel('Backscattering coefficient (dB)')
```
