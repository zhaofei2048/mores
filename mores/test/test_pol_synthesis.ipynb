```python
Mue = np.random.randn(4, 4)

print("POL: ref (calc)")
# linear
sigma0_VV, sigma0_VH, sigma0_HV, sigma0_HH = Mueller2sigma0VH(Mue)
print('sigma_VV: {} ({})'.format(4 * np.pi * Mue[0, 0], sigma0_VV))
print('sigma_VH: {} ({})'.format(4 * np.pi * Mue[0, 1], sigma0_VH))
print('sigma_HV: {} ({})'.format(4 * np.pi * Mue[1, 0], sigma0_HV))
print('sigma_HH: {} ({})'.format(4 * np.pi * Mue[1, 1], sigma0_HH))
# circular
sigma0_LL, sigma0_LR, sigma0_RL, sigma0_RR = Mueller2sigma0LR(Mue)
print('calc value: ')
print('sigma_LL: {} ({})'.format(4 * np.pi * 0.25 * (Mue[0, 0] + Mue[1, 1] + Mue[0, 1] + Mue[1, 0] - 2 * (Mue[3, 3] - (-Mue[0, 3] + 0.5 * Mue[3, 0] + 0.5 * Mue[3, 1] - Mue[1, 3]))), sigma0_LL))
print('sigma_LR: {} ({})'.format(4 * np.pi * 0.25 * (Mue[0, 0] + Mue[1, 1] + Mue[0, 1] + Mue[1, 0] + 2 * (Mue[3, 3] - (-Mue[0, 3] - 0.5 * Mue[3, 0] - 0.5 * Mue[3, 1] - Mue[1, 3]))), sigma0_LR))
# print('sigma_LR: {} ({})'.format(4 * np.pi * 0.25 * (Mue[0, 0] + Mue[1, 1] + Mue[0, 1] + Mue[1, 0] - 2*Mue[0, 3] + Mue[1, 0] + Mue[1, 1] - 2*Mue[1, 3] + Mue[3, 0] + Mue[3, 1] - 2*Mue[3, 3]), sigma0_LR))
print('sigma_RL: {} ({})'.format(4 * np.pi * 0.25 * (Mue[0, 0] + Mue[1, 1] + Mue[0, 1] + Mue[1, 0] + 2 * (Mue[3, 3] + (-Mue[0, 3] - 0.5 * Mue[3, 0] - 0.5 * Mue[3, 1] - Mue[1, 3]))), sigma0_RL))
print('sigma_RR: {} ({})'.format(4 * np.pi * 0.25 * (Mue[0, 0] + Mue[1, 1] + Mue[0, 1] + Mue[1, 0] - 2 * (Mue[3, 3] + (-Mue[0, 3] + 0.5 * Mue[3, 0] + 0.5 * Mue[3, 1] - Mue[1, 3]))), sigma0_RR))
```
